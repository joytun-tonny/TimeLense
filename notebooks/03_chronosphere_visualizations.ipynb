```python
# %% [markdown]
# # CHRONOSPHERE: Complete Visualization Suite
# ## Time-Travel Data Debugging in Action

# %% [markdown]
# ### 1. Setup

# %%
import sys
sys.path.append('../src')

import pandas as pd
import numpy as np
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots

# Import our custom modules
from visualization.chronosphere_viz import ChronosphereVisualizer
from core.data_processor import AnomalyDetector
from intelligence.insight_generator import InsightGenerator

print("✅ All modules loaded")

# %%
# Load processed data
print("Loading data...")
df = pd.read_csv('../data/processed/events_cleaned.csv', parse_dates=['event_time'])
hourly = pd.read_csv('../data/processed/hourly_aggregated.csv', parse_dates=['event_time'])
daily = pd.read_csv('../data/processed/daily_aggregated.csv', parse_dates=['date'])

print(f"✅ Data loaded: {len(df):,} events")
```

```python
# %% [markdown]
# ### 2. VISUALIZATION #1: Temporal Heartbeat

# %%
print("Creating Temporal Heartbeat visualization...")

viz = ChronosphereVisualizer(hourly, 'event_time')
fig = viz.temporal_heartbeat('total_events', title='Event Stream Heartbeat - Pulse Visualization')
fig.write_html('../assets/visualizations/viz1_temporal_heartbeat.html')
fig.show()

print("✅ Saved: viz1_temporal_heartbeat.html")

# %%
# Multiple metrics heartbeat
fig2 = viz.temporal_heartbeat('unique_users', title='User Activity Heartbeat')
fig2.show()
```

```python
# %% [markdown]
# ### 3. VISUALIZATION #2: Temporal Correlation Explorer

# %%
print("Creating Temporal Correlation Explorer...")

fig = viz.temporal_correlation_explorer('unique_users', 'total_events', n_frames=15)
fig.write_html('../assets/visualizations/viz2_correlation_explorer.html')
fig.show()

print("✅ Saved: viz2_correlation_explorer.html")
```

```python
# %% [markdown]
# ### 4. VISUALIZATION #3: 3D Phase Space with Ghost Trails

# %%
print("Creating 3D Phase Space...")

# Prepare user-level data for trajectories
user_daily = df.groupby(['user_id', df['event_time'].dt.date]).agg({
    'event_type': 'count',
    'product_id': 'nunique',
    'price': 'sum'
}).reset_index()
user_daily.columns = ['user_id', 'date', 'event_count', 'unique_products', 'total_spent']
user_daily['date'] = pd.to_datetime(user_daily['date'])

# Sample top users for performance
top_users = df['user_id'].value_counts().head(50).index
user_sample = user_daily[user_daily['user_id'].isin(top_users)]

print(f"Using {len(user_sample)} user-day combinations")

# %%
# Create 3D visualization
viz3d = ChronosphereVisualizer(user_sample, 'date')
fig = viz3d.phase_space_3d(
    'event_count',
    'unique_products', 
    'total_spent',
    color_col='event_count',
    entity_col='user_id',
    show_trails=True,
    max_entities=15
)
fig.write_html('../assets/visualizations/viz3_phase_space_3d.html')
fig.show()

print("✅ Saved: viz3_phase_space_3d.html")
```

```python
# %% [markdown]
# ### 5. VISUALIZATION #4: High-Dimensional Evolution (t-SNE)

# %%
print("Creating t-SNE Evolution...")

# Prepare feature matrix
hourly_features = hourly[['unique_users', 'total_events', 'unique_products', 'total_revenue']].fillna(0)

viz_hd = ChronosphereVisualizer(hourly, 'event_time')
fig = viz_hd.temporal_umap_evolution(
    ['unique_users', 'total_events', 'unique_products', 'total_revenue'],
    n_frames=12
)
fig.write_html('../assets/visualizations/viz4_tsne_evolution.html')
fig.show()

print("✅ Saved: viz4_tsne_evolution.html")
```

```python
# %% [markdown]
# ### 6. VISUALIZATION #5: The Chronosphere Interface (Main Feature)

# %%
print("Creating main Chronosphere interface...")

# Detect anomalies
detector = AnomalyDetector(hourly, 'event_time')
hourly_with_anomalies = detector.detect_zscore('total_events', threshold=2.5)
anomaly_indices = hourly_with_anomalies[hourly_with_anomalies['total_events_anomaly']].index.tolist()

print(f"Detected {len(anomaly_indices)} anomalies")

# %%
# Create Chronosphere rewind interface
viz_chrono = ChronosphereVisualizer(hourly, 'event_time')
fig = viz_chrono.chronosphere_rewind_interface(
    'total_events',
    anomaly_indices,
    window_hours=72
)
fig.write_html('../assets/visualizations/viz5_chronosphere_main.html')
fig.show()

print("✅ Saved: viz5_chronosphere_main.html")
```

```python
# %% [markdown]
# ### 7. VISUALIZATION #6: Parallel Coordinates

# %%
print("Creating Parallel Coordinates...")

# Sample data for better performance
hourly_sample = hourly.sample(n=min(500, len(hourly)), random_state=42)
viz_pc = ChronosphereVisualizer(hourly_sample, 'event_time')
fig = viz_pc.parallel_coordinates(
    ['unique_users', 'total_events', 'unique_products', 'total_revenue'],
    color_col='total_events'
)
fig.write_html('../assets/visualizations/viz6_parallel_coordinates.html')
fig.show()

print("✅ Saved: viz6_parallel_coordinates.html")
 
```

```python
# %% [markdown]
# ### 8. VISUALIZATION #7: Temporal Heatmap

# %%
print("Creating Temporal Heatmap...")

viz_heat = ChronosphereVisualizer(df, 'event_time')
fig = viz_heat.temporal_heatmap('price', x_groupby='hour', y_groupby='day_of_week')
fig.write_html('../assets/visualizations/viz7_temporal_heatmap.html')
fig.show()

print("✅ Saved: viz7_temporal_heatmap.html")
```

```python
# %% [markdown]
# ### 9. VISUALIZATION #8: Event Type Distribution Over Time

# %%
print("Creating event type evolution...")

# Group by date and event type
event_evolution = df.groupby([df['event_time'].dt.date, 'event_type']).size().reset_index(name='count')
event_evolution.columns = ['date', 'event_type', 'count']
event_evolution['date'] = pd.to_datetime(event_evolution['date'])

fig = px.area(event_evolution, x='date', y='count', color='event_type',
              title='Event Type Evolution Over Time',
              labels={'count': 'Number of Events', 'date': 'Date'},
              color_discrete_map={'view': '#00d4ff', 'cart': '#ff00ff', 'purchase': '#ffff00'})
fig.update_layout(template='plotly_dark', height=500)
fig.write_html('../assets/visualizations/viz8_event_evolution.html')
fig.show()

print("✅ Saved: viz8_event_evolution.html")
```

```python
# %% [markdown]
# ### 10. VISUALIZATION #9: Brand Performance Dashboard

# %%
print("Creating brand performance dashboard...")

# Top brands
top_brands = df[df['brand'] != 'Unknown'].groupby('brand').agg({
    'event_type': ['count', lambda x: (x == 'purchase').sum()],
    'price': 'sum'
}).reset_index()
top_brands.columns = ['brand', 'total_events', 'purchases', 'revenue']
top_brands['conversion_rate'] = (top_brands['purchases'] / top_brands['total_events']) * 100
top_brands = top_brands.nlargest(10, 'total_events')

# Create dashboard
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Top Brands by Events', 'Conversion Rates', 
                    'Revenue by Brand', 'Events vs Purchases'),
    specs=[[{'type': 'bar'}, {'type': 'bar'}],
           [{'type': 'bar'}, {'type': 'scatter'}]]
)

# Plot 1: Events
fig.add_trace(
    go.Bar(x=top_brands['brand'], y=top_brands['total_events'], marker_color='#00d4ff', name='Events'),
    row=1, col=1
)

# Plot 2: Conversion
fig.add_trace(
    go.Bar(x=top_brands['brand'], y=top_brands['conversion_rate'], marker_color='#ff00ff', name='Conv Rate'),
    row=1, col=2
)

# Plot 3: Revenue
fig.add_trace(
    go.Bar(x=top_brands['brand'], y=top_brands['revenue'], marker_color='#ffff00', name='Revenue'),
    row=2, col=1
)

# Plot 4: Scatter
fig.add_trace(
    go.Scatter(x=top_brands['total_events'], y=top_brands['purchases'], 
               mode='markers', marker=dict(size=10, color='#00d4ff'), name='Brands'),
    row=2, col=2
)

fig.update_layout(
    title_text="Brand Performance Dashboard",
    template='plotly_dark',
    height=800,
    showlegend=False
)
fig.update_xaxes(tickangle=-45)
fig.write_html('../assets/visualizations/viz9_brand_dashboard.html')
fig.show()

print("✅ Saved: viz9_brand_dashboard.html")
```

```python
# %% [markdown]
# ### 11. VISUALIZATION #10: User Journey Sankey Diagram

# %%
print("Creating user journey Sankey...")

# Sample users and their event sequences
sample_size = 1000
sample_sessions = df['user_session'].value_counts().head(sample_size).index
journey_df = df[df['user_session'].isin(sample_sessions)].sort_values(['user_session', 'event_time'])

# Create event sequences
journey_df['next_event'] = journey_df.groupby('user_session')['event_type'].shift(-1)
transitions = journey_df[journey_df['next_event'].notna()].groupby(['event_type', 'next_event']).size().reset_index(name='count')

# Create Sankey
event_types = list(set(transitions['event_type'].unique()) | set(transitions['next_event'].unique()))
event_map = {event: i for i, event in enumerate(event_types)}

fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=15,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=event_types,
        color=['#00d4ff', '#ff00ff', '#ffff00'][:len(event_types)]
    ),
    link=dict(
        source=[event_map[x] for x in transitions['event_type']],
        target=[event_map[x] for x in transitions['next_event']],
        value=transitions['count']
    )
)])

fig.update_layout(
    title="User Journey Flow (Sankey Diagram)",
    template='plotly_dark',
    height=600
)
fig.write_html('../assets/visualizations/viz10_user_journey.html')
fig.show()

print("✅ Saved: viz10_user_journey.html")
```

```python
# %% [markdown]
# ### 12. EXECUTIVE DASHBOARD

# %%
print("Creating executive dashboard...")

# Calculate KPIs
total_users = df['user_id'].nunique()
total_revenue = df[df['event_type'] == 'purchase']['price'].sum()
avg_order_value = df[df['event_type'] == 'purchase']['price'].mean()
conversion_rate = (df['event_type'] == 'purchase').sum() / (df['event_type'] == 'view').sum() * 100

# Create dashboard
fig = make_subplots(
    rows=3, cols=2,
    subplot_titles=('Daily Events', 'Daily Revenue',
                    'Hourly Pattern', 'Day of Week Pattern',
                    'Conversion Funnel', 'Key Metrics'),
    specs=[[{'type': 'scatter'}, {'type': 'scatter'}],
           [{'type': 'bar'}, {'type': 'bar'}],
           [{'type': 'funnel'}, {'type': 'indicator'}]],
    vertical_spacing=0.12
)

# Plot 1: Daily events
fig.add_trace(
    go.Scatter(x=daily['date'], y=daily['total_events'], mode='lines', 
               line=dict(color='#00d4ff', width=2), name='Events'),
    row=1, col=1
)

# Plot 2: Daily revenue
fig.add_trace(
    go.Scatter(x=daily['date'], y=daily['total_revenue'], mode='lines',
               line=dict(color='#ffff00', width=2), name='Revenue'),
    row=1, col=2
)

# Plot 3: Hourly
hourly_avg = hourly.groupby(hourly['event_time'].dt.hour)['total_events'].mean()
fig.add_trace(
    go.Bar(x=hourly_avg.index, y=hourly_avg.values, marker_color='#ff00ff', name='Hourly'),
    row=2, col=1
)

# Plot 4: Day of week
dow_avg = df.groupby('day_of_week').size()
fig.add_trace(
    go.Bar(x=['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'], y=dow_avg.values, 
           marker_color='#00d4ff', name='DOW'),
    row=2, col=2
)

# Plot 5: Funnel
funnel_values = [
    len(df[df['event_type'] == 'view']),
    len(df[df['event_type'] == 'cart']),
    len(df[df['event_type'] == 'purchase'])
]
fig.add_trace(
    go.Funnel(y=['View', 'Cart', 'Purchase'], x=funnel_values,
              marker=dict(color=['#00d4ff', '#ff00ff', '#ffff00'])),
    row=3, col=1
)

# Plot 6: KPI
fig.add_trace(
    go.Indicator(
        mode="number+delta",
        value=conversion_rate,
        title={'text': f"Conversion Rate<br><span style='font-size:0.8em'>Total Users: {total_users:,}</span>"},
        delta={'reference': 3.0, 'relative': False, 'suffix': '%'},
        number={'suffix': '%'},
        domain={'x': [0, 1], 'y': [0, 1]}
    ),
    row=3, col=2
)

fig.update_layout(
    title_text="🌐 CHRONOSPHERE Executive Dashboard",
    template='plotly_dark',
    height=1200,
    showlegend=False
)

fig.write_html('../assets/visualizations/chronosphere_dashboard.html')
fig.show()

print("✅ Saved: chronosphere_dashboard.html")
```
